```python

def car_rental():
    #Car type and daily rental rate
    car_type = {'economy': 50, 'standard': 75, 'luxury': 100}
    user_car_type = input("Hello! Please enter which car type you want to take for rent: Economy, Standard or Luxury: ")

    if user_car_type.lower() not in car_type:
        print('Invalid car type entered. Please enter a valid car type.')
        car_rental()
    else:
        rent = car_type[user_car_type.lower()]
        print(f'The car type you have chosen is {user_car_type.capitalize()} and the daily rate is ${rent}')

    #Rental duration
    try:
        rental_days = int(input('Enter how many days you want to take the car on rent: '))
        if rental_days <= 0:
            print('Please enter a valid number of days.')
            return
    except ValueError:
        print('Please enter numeric input for days.')
        return

    base_fee = rent * rental_days
    print(f'You have rented the car for {rental_days} days. Your base fee is ${base_fee}')

    #Membership discount
    membership = {'premium': 15, 'standard': 10, 'none': 0}
    member = input('Do you have any membership? (Premium, Standard, None): ')

    if member.lower() not in membership:
        print('Invalid membership type given.')
        return

    disc_rate = membership[member.lower()] / 100
    discount = base_fee * disc_rate
    discounted_fee = base_fee - discount
    print(f'As a {member.capitalize()} member, you got a discount of ${discount:.2f}. Fee after discount: ${discounted_fee:.2f}')

    #Optional Services
    insurance = input('Do you want Insurance (yes/no)? ').lower()
    gps = input('Do you want GPS (yes/no)? ').lower()

    insurance_fee = 20 * rental_days if insurance == 'yes' else 0
    gps_fee = 5 * rental_days if gps == 'yes' else 0
    optional_total = insurance_fee + gps_fee

    print(f'Optional services charges: Insurance = ${insurance_fee}, GPS = ${gps_fee}, Total = ${optional_total}')

    #Late Return Penalty
    try:
        late_days = int(input('Enter number of late return days (0 if none): '))
        if late_days < 0:
            print('Late days cannot be negative.')
            return
    except ValueError:
        print('Invalid input for late days.')
        return

    late_penalty = 30 * late_days if late_days > 0 else 0
    print(f'Late return penalty: ${late_penalty}')

    #Total rental cost
    total_fee = discounted_fee + optional_total + late_penalty

    print("\n----- Rental Summary -----")
    print(f"Car Type: {user_car_type.capitalize()}")
    print(f"Daily Rate: ${rent}")
    print(f"Rental Days: {rental_days}")
    print(f"Base Fee: ${base_fee}")
    print(f"Membership Discount: -${discount:.2f}")
    print(f"Fee After Discount: ${discounted_fee:.2f}")
    print(f"Insurance Fee: ${insurance_fee}")
    print(f"GPS Fee: ${gps_fee}")
    print(f"Late Penalty: ${late_penalty}")
    print(f"Total Rental Fee: ${total_fee:.2f}")
```

```python
car_rental()
```
